# ¿Quiénes participan en las elecciones en México? Hallazgos de 2009 a 2024
**Hackathon INE 2026 · Reto 1: perfiles ciudadanos de participación**

Este documento resume lo que muestran los conteos de participación del INE en seis elecciones federales. Cada hallazgo indica **qué se ve**, **qué técnica se usó** y **por qué**, y separa lo que los datos respaldan de lo que no. Los resultados describen grupos y territorios, no personas.

**Lo esencial:** (1) la participación depende sobre todo de la edad y del tipo de elección; (2) las mujeres participan más que los hombres, salvo después de los 70 años; (3) los territorios difieren mucho, pero no forman tipos separados: forman un gradiente; (4) agrupar los estados en 3 regiones (norte, centro, sur) no explica el efecto territorial, y en 5 regiones sí explica la mitad; (5) dentro de cada región hay municipios muy distintos.

## 1. Los datos y sus variables
Se usaron los **Conteos Censales de Participación Ciudadana** del INE para las elecciones de 2009, 2012, 2015, 2018, 2021 y 2024 (54.9 millones de filas, 32 entidades). Cada fila es un grupo: una sección electoral, un sexo y una edad en años cumplidos, con cuatro conteos.

| Variable | Qué significa |
| --- | --- |
| Lista nominal (LN) | Personas registradas para votar en ese grupo |
| Con marca de voto (SV) | Registros donde consta que votaron |
| Sin marca de voto (NV) | Registros donde no consta voto |
| No especificado (NS) | No se sabe: cuadernillos sin marcas, no disponibles o solicitados por una autoridad |
| Sexo, edad, tipo de sección | Hombres, mujeres (y, solo en 2024, 105 registros no binarios que se excluyen); edad de 18 a 140 años; sección urbana, mixta o rural |
| Entidad y municipio | Ubicación según la credencial de elector |

**Lo que estos datos no incluyen** (según el diccionario del INE): votos en casillas especiales y votos desde el extranjero. Tampoco son resultados por partido. No hay escolaridad ni ingreso.

**Calidad:** en las seis elecciones las partes suman exactamente el total (LN = SV + NV + NS), sin duplicados ni valores vacíos. Se observó que hay 91,629 filas con edades mayores a 100 años (hasta 140), que son 0.04% o menos de la lista; se conservaron y no cambian los resultados. La región (norte, centro, sur) es una propuesta propia de este análisis, no una división oficial.

## 2. Cómo se mide la participación
**Participación = personas con marca de voto ÷ (con marca + sin marca).** Los registros "no especificados" se dejan fuera del denominador porque no se sabe si votaron.

Si se divide entre la lista nominal completa, la participación baja: 55.8% en vez de 59.7% en 2024. La diferencia depende de cuántos registros no especificados hay (0.2% en 2012; 6.4% en 2024; 14.9% en Chihuahua y Campeche en 2024). Por eso las comparaciones entre estados deben leerse con esa cautela.

Para un grupo (por ejemplo, todas las mujeres de un estado) siempre se **suman los conteos** y después se calcula el porcentaje; nunca se promedian porcentajes.

Cuando se comparan grupos, se pregunta si la diferencia se debe a que tienen **edades distintas**. Para ello se usa la **estandarización directa**: se calcula qué participación tendría cada grupo si todos tuvieran la misma distribución de edad (y de sexo, según el caso).

## 3. Hallazgos más representativos

### 3.1 La participación sube en elecciones presidenciales y cae en las intermedias
**Qué se ve.** 62.1% en 2012, 62.4% en 2018 y 59.7% en 2024, contra 44.1% en 2009, 47.1% en 2015 y 51.8% en 2021.
**Técnica y por qué.** Tasa nacional sumando conteos, con dos denominadores, para ver cuánto depende la cifra de cómo se mida. La clasificación de las elecciones como presidenciales o intermedias es conocimiento externo, no sale de los datos.

![Participación nacional por elección](outputs/graficas/01_participacion_nacional.png)

### 3.2 Las mujeres participan más que los hombres, y no es por la edad
**Qué se ve.** La diferencia mujeres − hombres va de 6.8 a 9.4 puntos. Al comparar con la misma estructura de edad va de 6.3 a 9.3 puntos; la de 2024 es la más alta.
**Técnica y por qué.** Estandarización directa por edad, para descartar que las mujeres participen más solo porque haya otra distribución de edades entre ellas y ellos.

![Brecha entre mujeres y hombres](outputs/graficas/02_brecha_sexo.png)

### 3.3 El promedio oculta que, después de los 70 años, votan más los hombres
**Qué se ve.** En 2024 las mujeres votan 13.7 puntos más entre los 30-39 años, pero entre los de 70 o más votan 5.4 puntos menos que los hombres. Ocurre en las seis elecciones.
**Técnica y por qué.** Brecha por grupo de edad y elección, porque una sola cifra nacional mezcla grupos que se comportan de forma contraria. Los datos muestran el patrón, no sus razones.

![Brecha por sexo según la edad](outputs/graficas/03_brecha_sexo_edad.png)

### 3.4 De 2018 a 2024 la participación bajó antes de los 60 años y subió después
**Qué se ve.** Cambió −4.4 puntos entre los 18-24 años, −5.6 entre 25-29, −6.0 entre 30-39, −4.2 entre 40-49, −2.2 entre 50-59, +1.5 entre 60-69 y +4.5 en mayores de 70; el total cambió −2.7.
**Técnica y por qué.** Diferencia entre las dos elecciones presidenciales comparables, sumando conteos por edad.

![Cambio por edad entre 2018 y 2024](outputs/graficas/04_cambio_edad_2018_2024.png)

### 3.5 Entre estados hay hasta 24 puntos de diferencia, y la edad y el sexo casi no lo explican
**Qué se ve.** En 2024 va de 46.7% (Baja California) a 71.1% (Yucatán). Al igualar la composición de sexo y edad, el orden de los estados casi no cambia: solo 1 de 32 se mueve más de 3 lugares.
**Técnica y por qué.** Estandarización directa por sexo y edad, para saber si la diferencia entre estados viene de que unos tengan más jóvenes o mayores que otros.

![Participación por estado en 2024](outputs/graficas/05_entidades_2024.png)

### 3.6 No hay una región que siempre participe más
**Qué se ve.** Con la misma composición de sexo y edad, el Sur lidera en 2012, 2015, 2018 y 2021 y el Centro en 2009 y 2024. El Norte es la más baja de 2012 a 2024.
**Técnica y por qué.** Tasas por región (propuesta propia de este análisis) con estandarización directa, repetidas en cada elección.

![Participación por región](outputs/graficas/06_regiones_tiempo.png)

### 3.7 La ventaja de las secciones rurales prácticamente desapareció
**Qué se ve.** Con sexo y edad igualados, las secciones rurales superaban a las urbanas por 10.0 puntos en 2015 y por 0.8 en 2024.
**Técnica y por qué.** Estandarización directa por sexo y edad, para comparar tipos de sección sin que cambie la composición. En el modelo estadístico (sección 4), la diferencia de 2024 queda entre −2.0 y +3.8 puntos con 95% de confianza.

![Participación por tipo de sección](outputs/graficas/07_tipo_seccion.png)

### 3.8 La edad es el factor que más se asocia con la participación
**Qué se ve.** Por sí sola, la edad explica 33% de las diferencias entre grupos; la elección, 26%; el estado, 11%; el sexo, 8%; y el tipo de sección, 1.5%.
**Técnica y por qué.** Un **modelo de regresión logística** para conteos, que añade los factores de uno en uno y mide cuánto aporta cada uno. Es el tipo de modelo apropiado cuando lo que se cuenta es cuántas personas votaron de un total.

![Cuánto explica cada factor](outputs/graficas/11_factores_modelo.png)

### 3.9 En puntos porcentuales: ¿cuánta diferencia hay entre grupos?
**Qué se ve.** Con todo lo demás igual, las mujeres participan 7.8 puntos más que los hombres; las personas de 60-69 años, 18.0 puntos más que las de 30-39; las de 25-29, 6.4 menos. 2024 estuvo 3.4 puntos por debajo de 2018 aun igualando la composición.
**Técnica y por qué.** Los coeficientes del modelo se traducen a diferencias en puntos porcentuales (efectos marginales promedio), con intervalos de confianza por remuestreo de estados, porque los coeficientes originales (razones de momios) son difíciles de interpretar.

![Diferencias en puntos porcentuales](outputs/graficas/12_efectos_puntos.png)

### 3.10 Los estados no forman tipos: forman un gradiente
**Qué se ve.** Se probó agrupar los 32 estados por su historia de participación con cuatro métodos y varios números de grupos. Los grupos resultaron no más nítidos que los que aparecen al azar: datos simulados sin grupos separan igual o mejor que los reales en 67% de los casos.
**Técnica y por qué.** Agrupamiento por distancia (K-means, Ward) y por densidad (DBSCAN, HDBSCAN). Se comparó contra datos simulados sin grupos y contra un simple corte por nivel promedio (que explica 41% frente a 42% del mejor agrupamiento). Es una prueba de escepticismo: un algoritmo siempre entrega grupos, aunque no existan.

![Agrupamiento frente a datos simulados](outputs/graficas/08_agrupamiento_vs_nulo.png)

### 3.11 Lo que sí hay son dos ejes continuos
**Qué se ve.** La historia de cada estado se resume bien con dos ejes: su nivel general de participación (64% de la variación) y si es relativamente más alto en 2009-2015 o en 2018-2024 (20%). No hay islas de estados parecidos.
**Técnica y por qué.** Análisis de componentes principales, que resume varias columnas (las seis elecciones) en pocas dimensiones, como alternativa continua cuando los grupos no son estables.

![Mapa de estados por trayectoria](outputs/graficas/10_entidades_continuo_pca.png)

### 3.12 Tres regiones no bastan; cinco recogen la mitad del efecto de estado
**Qué se ve.** Con tres regiones (norte, centro, sur) se recoge 21% de la diferencia entre estados, lo mismo que lograría una agrupación al azar (solo 5% de las agrupaciones al azar supera 24%). Con cinco regiones se recoge 54%, bastante más que el azar (33%).
**Técnica y por qué.** Se midió cuánto del efecto de "estado" del modelo recoge una región. Se comparó contra 200 agrupaciones al azar del mismo tamaño, para saber si el resultado es mérito de la agrupación o del número de grupos. Ambas regionalizaciones son propuestas propias, no oficiales.

![Captura del efecto de estado por región](outputs/graficas/13_regiones_captura.png)

### 3.13 Dentro de cada región, los municipios son muy distintos
**Qué se ve.** En 2024, el 80% central de los municipios de una región abarca entre 18 y 36 puntos de participación, según la región. Con tres regiones, 30% de los municipios del Sur (la región con menor mediana) supera la mediana del Centro (la de mayor mediana).
**Técnica y por qué.** Participación por municipio ajustada por sexo y edad (estandarización indirecta, porque los municipios pequeños no permiten la directa) y descomposición de la varianza. Entre 42% y 67% de la variación total ocurre entre municipios de un mismo estado, y el azar de conteos pequeños aporta menos de 0.2%.

![Dispersión de municipios dentro de cada región](outputs/graficas/14_municipios_dentro_region.png)

## 4. Técnicas usadas y por qué

| Técnica | Para qué se usó |
| --- | --- |
| Verificación aritmética, de duplicados y de extremos | Saber si los datos se pueden usar; se reporta sin corregir |
| Suma de conteos | Evitar el error de promediar porcentajes de grupos de distinto tamaño |
| Estandarización directa | Separar la diferencia entre grupos del efecto de que tengan edades o sexos distintos |
| Agrupamiento (distancia y densidad), con validación | Ver si hay tipos naturales de estado; se contrastó con datos simulados, retiro de estados y elecciones, y un corte simple |
| Componentes principales | Resumir la trayectoria de cada estado sin forzar grupos |
| Regresión logística para conteos, con errores robustos por estado | Medir cuánto aporta cada factor y las diferencias en puntos porcentuales |
| Comparación contra particiones al azar | Saber si una región "recoge" el territorio o solo suma grupos |
| Descomposición de varianza municipal | Ver cuánta variación queda dentro de cada región |

## 5. Conclusiones y límites

**Lo que los datos respaldan**
- La participación varía sobre todo por edad y por tipo de elección; el sexo y el territorio importan, y el tipo de sección importa poco.
- Las mujeres participan más que los hombres hasta los 50-59 años en las seis elecciones (y hasta los 60-69 desde 2012); a partir de los 70 ocurre al revés.
- De 2018 a 2024, bajó la participación de las personas menores de 60 años y subió la de mayores.
- Los estados forman un gradiente, no tipos. Las tres regiones norte-centro-sur no recogen el efecto de estado más que el azar; cinco regiones recogen la mitad. Dentro de las regiones hay mucha variación municipal.

**Lo que los datos no respaldan**
- **Causas y motivos.** Nada de esto explica *por qué* participan o no. Son asociaciones entre grupos.
- **Individuos.** No se sigue a personas entre elecciones; "las personas de 18-24 años" son grupos distintos en cada elección.
- **Votos por partido o candidatura**, casillas especiales y voto en el extranjero (no están en la fuente).

**Límites que conviene tener presentes**
- Las dos regionalizaciones y los grupos de edad son decisiones de este análisis; cambiarlas puede cambiar cifras (se pueden modificar en `scripts/config.py`).
- No se pudo comparar contra el dataset maestro local del equipo (no estaba disponible); todo se reconstruyó desde los ZIP originales.
- Los registros "no especificados" son muy distintos entre estados y años (0.2% a 14.9%); eso afecta las comparaciones.
- Con solo 32 estados, las pruebas por estado tienen poco poder y los intervalos por estado son aproximados.
